**Step-4 Add Multiple columns**

```python
names=["Ramesh","Suresh","Sathish"]
age=[20,22,24]
pd.DataFrame(zip(names,age))
```

```python
#adding columns

names=["Ramesh","Suresh","Sathish"]
age=[20,22,24]
pd.DataFrame(zip(names,age),
            columns=["Name","Age"])
```

```python
#Adding indexes

names=["Ramesh","Suresh","Sathish"]
age=[20,22,24]
pd.DataFrame(zip(names,age),
            columns=["Name","Age"],
            index=['A','B','C'])
```

- Incase of labels, we need to mention axis

**Step-12 - loc,iloc**
- Loc uses for categorical variables
- iloc uses index(numbers)

```python
#using loc 

df1.loc[3:7,['square_value','cube_value']]   #loc takes variables --
```

```python
df1.iloc[3:7,1:3]    #iloc --- i means index--- so we are passing numbers

#3 to 7-- rows (3 to 6)
#1:3 -- column (1 , 2)
```

```python
cols=['square_value','cube_value']
df1.loc[3:7,cols]
```

- Single Bracket -- 1 Dimensional
- 2 Brackets -- 2 Dimensional

```python
df1.iloc[[2,3,4]]

#rows -- 2,3,4 
#columns -- since nothing is specified, all columns will be displayed
#Data frame --2 d
```

```python
df1.iloc[[2]]

#2nd row alone is printed
#2 square brackets -- 
#data frame -- 2d
```

```python
df1.iloc[2]

#series -- single square bracket -- 1D
```

```python
df1.iloc[[2,3,4],[0,2]]

#0,2 -- columns

#2 D
```

```python
df1.iloc[2,0]

#Series --- 1D
```

```python
df1.iloc[:,0]

#Series -- 1D
# All rows, 1 column
```

```python
df1.iloc[2,:]

#2nd row, 1 column
#Series - 1D
```

```python
df1.iloc[[2],:]

#2nd row alone
#Data frame- 2D
```

```python
df1.iloc[2:6,0]

#Series- 1d
```

```python
df1.iloc[2:6,[0]]
```
