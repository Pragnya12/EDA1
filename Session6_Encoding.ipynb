**LabelEncoder vs OnehotEncoder**

- Generally we dont have a relation between input columns

- we need both input columns which are not correlated to each other

- we need both input columns which are independent (X,Y) to each other

- We need both columns which are perpendicular to each other

- We need both columns which are orthogonal to each other

- We need both columns which are 90 degrees phase shift to each other

- OneHotEncoder will achieve above..

- But OneHotEncoder creates new columns

    - Columns are increased means dimensions are increased
 
    - More dimension data need more processing time
 
    - More dimension data need more infra
 
    - This is called **Curse of Dimensionality**
