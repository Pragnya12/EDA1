```python
cert_count,deni_count=[],[]
for i in visa_df['continent'].unique():
    con1=visa_df['continent']==i
    con2=visa_df['case_status']=='Certified'
    con3=visa_df['case_status']=='Denied'
    cert_con=con1&con2
    Denied_con=con1&con3
    count_certified=len(visa_df[cert_con])
    count_denied=len(visa_df[Denied_con])
    cert_count.append(count_certified)
    deni_count.append(count_denied)
```

```python
cert_count,deni_count
```

```python
data=zip(cert_count,deni_count)
cols=visa_df['case_status'].unique()
idx=visa_df['continent'].unique()
pd.DataFrame(data,index=idx,columns=cols)
```

```python
#in above Denied came frst.. to avoid that reverse cols usng list

data=zip(cert_count,deni_count)
cols=visa_df['case_status'].unique()[::-1]
idx=visa_df['continent'].unique()
pd.DataFrame(data,index=idx,columns=cols)
```
